# 01. Exploratory Data Analysis (EDA) - Dữ liệu thô
**Môn học:** IS217 - Kho dữ liệu và OLAP
**Mục tiêu:** Khám phá cấu trúc các file dữ liệu thô trong `data/01_raw`, kiểm tra giá trị thiếu (Missing values), phân bố các biến số và chuẩn bị logic làm sạch.

```python
import pandas as pd
import numpy as np
from pathlib import Path

RAW_DIR = Path('../data/01_raw')
print('Danh sach file:', [f.name for f in RAW_DIR.glob('*.csv')])
```

## 1. Khám phá file Quốc gia (`dim_quocgia.csv`)

```python
df_quocgia = pd.read_csv(RAW_DIR / 'dim_quocgia.csv')
print('Shape:', df_quocgia.shape)
df_quocgia.info()
df_quocgia.head()
```

## 2. Khám phá file Kinh tế - Xã hội (`fact_kinhtexahoi.csv`)

```python
df_econ = pd.read_csv(RAW_DIR / 'fact_kinhtexahoi.csv')
print('Shape:', df_econ.shape)
print('So dong thieu GDP:', df_econ['gdp_per_capita_usd'].isnull().sum())
df_econ.describe()
```

## 3. Khám phá file GBD Mental Health

```python
df_gbd = pd.read_csv(RAW_DIR / 'GBD_mental_health_20-54_male_female_2013-2023.csv')
print('Shape:', df_gbd.shape)
print('Danh muc benh:', df_gbd['cause_name'].unique())
print('Thước đo:', df_gbd['measure_name'].unique())
print('Đơn vị:', df_gbd['metric_name'].unique())
```
